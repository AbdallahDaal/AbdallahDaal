```python
from imblearn.over_sampling import SMOTE

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import accuracy_score, precision_score, f1_score, recall_score
from sklearn.metrics import average_precision_score
from sklearn.metrics import precision_recall_curve
from sklearn.metrics import classification_report
```

```python
#create features and target datas
X = df.drop('Class', axis=1)
y = df['Class']

#split train_test ratio
X_train, X_test, y_train, y_test = train_test_split(X, y, stratify=y, random_state=42)

#imbalance solutioning with smote sampler
smoter = SMOTE(random_state=42)
X_train_sm, y_train_sm = smoter.fit_resample(X_train, y_train)
```

```python
#comparing the training set and the over sampled training set
print(X_train_sm.shape)
print(X_train.shape)
```

```python
plt.figure(figsize=(6,6))
sns.countplot(data=X_train_sm, x=y_train_sm)
plt.show()
```

```python
#scaling? yes, recommended for LogisticRegression model/
scaler = StandardScaler()
X_train_scale = scaler.fit_transform(X_train_sm)
X_test_scale = scaler.transform(X_test)
```

```python
X_train_scale.shape
```

```python
#
dtclf = DecisionTreeClassifier(random_state=42, max_depth=5)
log_reg = LogisticRegression(random_state=42, max_iter=50)
nbc = GaussianNB()
rfc = RandomForestClassifier()
xgb = XGBClassifier()
```

```python
clf = {
'Logistic Regression': log_reg,
'Decision Trees': dtclf,
'GausianNB' : nbc,
'Random Forest' : rfc,
'XGBoost' : xgb
}

evaluations = []

for classifier_name, classifier in clf.items():
    evals = {}
    y_pred = classifier.fit(X_train_scale, y_train_sm).predict(X_test_scale)
    evals['Accuracy'] = accuracy_score(y_test, y_pred)
    evals['Precision'] = precision_score(y_test, y_pred, pos_label=1)
    evals['Recall'] = recall_score(y_test, y_pred, pos_label=1)
    evals['F1'] = f1_score(y_test, y_pred, pos_label=1)
    evals['AUPRC'] = average_precision_score(y_test, y_pred)
    evaluations.append(evals)

metrics = pd.DataFrame(evaluations, index=['Logistic Regression', 'Decision Trees', 'GaussianNB', 'Random Forest', 'XGboost'])
metrics
```
